```python
# -----------------------------------------
# FINAL LEAKAGE-SAFE FEATURE ENGINEERING
# -----------------------------------------

# Start from the cleaned dataset
df_model = df.copy()

# Feature 1: whether the customer has a payment delay
df_model["has_payment_delay"] = (
    df_model["PAY_0"] > 0
).astype(int)

# Feature 2: payment-to-bill ratio
df_model["payment_to_bill_ratio"] = np.where(
    df_model["BILL_AMT1"] > 0,
    df_model["PAY_AMT1"] / df_model["BILL_AMT1"],
    0
)

# Feature 3: average repayment status
pay_cols = [
    "PAY_0", "PAY_2", "PAY_3",
    "PAY_4", "PAY_5", "PAY_6"
]

df_model["avg_repayment_status"] = df_model[pay_cols].mean(axis=1)

# Separate features and target
X_raw = df_model.drop(
    columns=["default payment next month"]
)

y_final = df_model["default payment next month"]

# Remove the raw ratio because of extreme outliers
X_raw = X_raw.drop(columns=["payment_to_bill_ratio"])

# Remove duplicate feature rows AND keep target aligned
X_raw = X_raw.loc[~X_raw.duplicated()].copy()
y_final = y_final.loc[X_raw.index].copy()

print("Final dataset:", X_raw.shape)
print("Target:", y_final.shape)
print("Missing values:", X_raw.isnull().sum().sum())
print("Duplicate rows:", X_raw.duplicated().sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_raw,
    y_final,
    test_size=0.2,
    random_state=42,
    stratify=y_final
)

print("Train:", X_train.shape)
print("Test :", X_test.shape)
```

```python
ratio_cap_train = X_train["PAY_AMT1"].div(
    X_train["BILL_AMT1"].replace(0, np.nan)
).fillna(0).quantile(0.99)

print("Training-only ratio cap:", ratio_cap_train)
```

```python
X_train["payment_to_bill_ratio_capped"] = np.where(
    X_train["BILL_AMT1"] > 0,
    (X_train["PAY_AMT1"] / X_train["BILL_AMT1"]).clip(
        upper=ratio_cap_train
    ),
    0
)

X_test["payment_to_bill_ratio_capped"] = np.where(
    X_test["BILL_AMT1"] > 0,
    (X_test["PAY_AMT1"] / X_test["BILL_AMT1"]).clip(
        upper=ratio_cap_train
    ),
    0
)

print("Train:", X_train.shape)
print("Test :", X_test.shape)
print("Train max ratio:", X_train["payment_to_bill_ratio_capped"].max())
print("Test max ratio:", X_test["payment_to_bill_ratio_capped"].max())
```

```python
print(X_train.columns.tolist())
```

```python
# Remove old ratio features
X_train = X_train.drop(
    columns=["payment_to_bill_ratio_capped_final"]
)

X_test = X_test.drop(
    columns=["payment_to_bill_ratio_capped_final"]
)

print("Train:", X_train.shape)
print("Test :", X_test.shape)
print("Columns:", X_train.columns.tolist())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

final_models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Balanced Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

final_results = []

for name, model in final_models.items():

    if "Logistic" in name:
        model.fit(X_train_scaled, y_train)
        predictions = model.predict(X_test_scaled)
        probabilities = model.predict_proba(X_test_scaled)[:, 1]

    else:
        model.fit(X_train, y_train)
        predictions = model.predict(X_test)
        probabilities = model.predict_proba(X_test)[:, 1]

    final_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1": f1_score(y_test, predictions),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    })

final_results_df = pd.DataFrame(final_results)

print(final_results_df.round(4))
```

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [None, 10, 15, 20, 25],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", "log2"]
}

rf_tuned_final = RandomizedSearchCV(
    estimator=RandomForestClassifier(
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),
    param_distributions=param_grid,
    n_iter=15,
    scoring="roc_auc",
    cv=5,
    random_state=42,
    n_jobs=-1
)

rf_tuned_final.fit(X_train, y_train)

print("Best parameters:")
print(rf_tuned_final.best_params_)

print("\nBest CV ROC-AUC:")
print(rf_tuned_final.best_score_)
```

```python
best_rf_final = rf_tuned_final.best_estimator_

y_pred_final = best_rf_final.predict(X_test)
y_prob_final = best_rf_final.predict_proba(X_test)[:, 1]

print("Accuracy :", accuracy_score(y_test, y_pred_final))
print("Precision:", precision_score(y_test, y_pred_final))
print("Recall   :", recall_score(y_test, y_pred_final))
print("F1 Score :", f1_score(y_test, y_pred_final))
print("ROC-AUC  :", roc_auc_score(y_test, y_prob_final))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_final))
```
